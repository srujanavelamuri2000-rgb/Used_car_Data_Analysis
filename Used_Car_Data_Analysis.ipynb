```python
#removing the column from this dataset
data=df.drop(['AdditionInfo'],axis=1)
data.info()
```

```python
from matplotlib import pyplot as plt
```

```python
import numpy as np
```

```python
#removing rupee symbol and changing into integer
data['AskPrice']=pd.to_numeric(data['AskPrice'].astype(str).str.replace('₹','',regex=False).str.replace(',','',regex=False),errors='coerce')
```

```python
data['kmDriven']=pd.to_numeric(data['kmDriven'].astype(str).str.replace(' km','',regex=False).str.replace(',','',regex=False),errors='coerce')
```

```python
data[['AskPrice','kmDriven']].describe()
```

```python
#missing value percentage
(data.isnull().sum()/len(data))*100
```

```python
#checking duplicates
data.duplicated().sum()
```

```python
data[data.duplicated()]
```

```python
data.drop_duplicates().shape
```

```python
data[data.duplicated()].head(10)
```

```python
#cleaning the duplicates
clean_data=data.drop_duplicates()
```

```python
#data after cleaning the duplicates
clean_data.shape
```

```python
#checking the missing values count after dropping the duplicates
clean_data.isnull().sum()
```

```python
clean_data.describe()
```

```python
fig,axarr=plt.subplots(3,2,figsize=(12,18))
clean_data.groupby('Brand')['AskPrice'].mean().plot.bar(ax=axarr[0][0],fontsize=8)
clean_data.groupby('Year')['AskPrice'].mean().plot.bar(ax=axarr[0][1],fontsize=8)
clean_data.groupby('Age')['AskPrice'].mean().plot.bar(ax=axarr[1][0],fontsize=8)
clean_data.groupby('Transmission')['AskPrice'].mean().plot.bar(ax=axarr[1][1],fontsize=8)
clean_data.groupby('FuelType')['AskPrice'].mean().plot.bar(ax=axarr[2][0],fontsize=8)
clean_data.groupby('Owner')['AskPrice'].mean().plot.bar(ax=axarr[2][1],fontsize=8)
plt.subplots_adjust(hspace=1.0)
plt.subplots_adjust(wspace=0.5)
```

# Conclusion

The Used Car Data Analysis project helped in understanding and exploring a dataset containing information about 9,582 used car listings.

During data cleaning, missing values, duplicate records and incorrect data types were examined. The asking price and mileage columns were converted into numerical formats. A total of 801 duplicate records were identified and removed, resulting in 8,781 records.

The analysis showed considerable variation in used car asking prices. The median asking price was ₹5.9 lakh, while the average asking price was approximately ₹10.4 lakh.

Overall, this project provided practical experience in Python, Pandas, data cleaning and exploratory data analysis. It also helped develop an understanding of how data analysis can be used to explore used car market trends.
